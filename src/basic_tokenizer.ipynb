```python
from tensorflow import keras
text = "LSTM e GRU são tipos de células em redes neurais recorrentes"
tokenizer = keras.preprocessing.text.Tokenizer(char_level=True) #IDs para caracteres
tokenizer.fit_on_texts([text])
print(f'Encontrei {len(tokenizer.word_index)} caracteres diferentes')
print(f'Ids: {tokenizer.word_index}')
```

```python
tokenizer.texts_to_sequences(['Gente'])

```

```python
tokenizer = keras.preprocessing.text.Tokenizer(char_level=False) #IDs para palavras
tokenizer.fit_on_texts([text])
print(f'Encontrei {len(tokenizer.word_index)} palavras diferentes')
print(f'Ids: {tokenizer.word_index}')
```

```python
tokenizer.texts_to_sequences(['redes recorrentes'])
```
